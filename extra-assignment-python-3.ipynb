```python
# Catagory the people by their age like children, adult, citizen, senior citizen..
people_age=int(input("Enter the age of the people : "))
if(people_age < 12):
    print("Children")
elif(people_age > 13 and people_age < 30):
    print("Adult")
elif(people_age > 22 and people_age < 55):
    print("Citizen")
else :
    print("Senior Citizen")
```
