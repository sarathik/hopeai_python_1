```python
#Find the BMI Category
#Get the Weight and Height from the user and calculate the BMI
#Based on the BMI Index value, find the category
#BMI=(weight/(height*height))*100
```

```python
print("Please enter the weight: ")
bodyWeight=int(input())
print("Please enter the height in cm: ")
bodyHeight=int(input())
bmiValue=bodyWeight/((bodyHeight*bodyHeight)/10000)
print("Your BMI is : ",bmiValue)
if(bmiValue < 18.5):
    print("Underweight")
elif(bmiValue > 18 and bmiValue < 25):
    print("Healthy Weight")
elif(bmiValue > 25 and bmiValue < 30):
    print("Overweight")
elif(bmiValue > 30 and bmiValue < 35):
    print("Obesity (Class 1)")
elif(bmiValue > 35 and bmiValue < 40):
    print("Obesity (Class 2)")
else:
    print("Severe Obesity (Class 3)")
```
